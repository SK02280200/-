# 03 高维定积分 high-dimensional definite integral

## 1. 算法介绍



高维定积分的随机模拟，通常使用 **Monte Carlo（蒙特卡洛）方法** 来近似计算多重积分。

对于一维或二维积分，传统的数值积分方法通常比较有效。但是当积分维数很高时，例如 10 维、50 维甚至更高维，传统网格划分法需要计算大量网格点，计算量会迅速增加，这就是所谓的“维数灾难”。

蒙特卡洛积分的基本思想是：不对积分区域进行大量网格划分，而是在积分区域中随机生成大量样本点，通过这些随机点上函数值的平均值来估计积分。

设需要计算的高维积分为：

$$
I=\int_D f(\mathbf{x})\,d\mathbf{x}
$$

其中：

$$
\mathbf{x}=(x_1,x_2,\dots,x_d)
$$

如果积分区域 $D$ 是一个 $d$ 维长方体，则可以在区域 $D$ 内均匀随机生成 $N$ 个样本点：

$$
\mathbf{X}_1,\mathbf{X}_2,\dots,\mathbf{X}_N
$$

计算这些随机点上的函数值：

$$
f(\mathbf{X}_1),f(\mathbf{X}_2),\dots,f(\mathbf{X}_N)
$$

再计算函数值的平均值：

$$
\bar f=\frac{1}{N}\sum_{i=1}^{N}f(\mathbf{X}_i)
$$

最后用：

$$
I\approx V(D)\bar f
$$

来估计高维定积分，其中 $V(D)$ 表示积分区域 $D$ 的体积。

## 2. 数学原理



设积分区域为一个 $d$ 维长方体：

$$
D=[a_1,b_1]\times[a_2,b_2]\times\cdots\times[a_d,b_d]
$$

需要计算：

$$
I=
\int_{a_1}^{b_1}
\int_{a_2}^{b_2}
\cdots
\int_{a_d}^{b_d}
f(x_1,x_2,\dots,x_d)
\,dx_d\cdots dx_2dx_1
$$

该积分区域的体积为：

$$
V(D)=\prod_{j=1}^{d}(b_j-a_j)
$$

在区域 $D$ 中随机生成一个向量：

$$
\mathbf{X}=(X_1,X_2,\dots,X_d)
$$

其中各分量满足：

$$
X_j\sim U(a_j,b_j)
$$

并且各个随机变量相互独立。

由于 $\mathbf{X}$ 在积分区域 $D$ 上服从均匀分布，所以其联合概率密度函数为：

$$
p(\mathbf{x})=\frac{1}{V(D)}
$$

函数 $f(\mathbf{X})$ 的数学期望为：

$$
E[f(\mathbf{X})]
=
\int_D f(\mathbf{x})p(\mathbf{x})\,d\mathbf{x}
$$

代入概率密度函数：

$$
E[f(\mathbf{X})]
=
\frac{1}{V(D)}
\int_D f(\mathbf{x})\,d\mathbf{x}
$$

由于：

$$
I=\int_D f(\mathbf{x})\,d\mathbf{x}
$$

所以：

$$
E[f(\mathbf{X})]
=
\frac{I}{V(D)}
$$

因此：

$$
I=V(D)E[f(\mathbf{X})]
$$

这就是高维蒙特卡洛积分的理论基础。

由于通常无法直接得到：

$$
E[f(\mathbf{X})]
$$

所以随机生成 $N$ 个相互独立的样本：

$$
\mathbf{X}_1,\mathbf{X}_2,\dots,\mathbf{X}_N
$$

根据大数定律，当 $N$ 足够大时：

$$
\frac{1}{N}
\sum_{i=1}^{N}
f(\mathbf{X}_i)
\longrightarrow
E[f(\mathbf{X})]
$$

因此可以近似认为：

$$
E[f(\mathbf{X})]
\approx
\frac{1}{N}
\sum_{i=1}^{N}
f(\mathbf{X}_i)
$$

代入积分公式：

$$
I=V(D)E[f(\mathbf{X})]
$$

得到：

$$
I
\approx
V(D)
\frac{1}{N}
\sum_{i=1}^{N}
f(\mathbf{X}_i)
$$

因此蒙特卡洛高维积分的估计公式为：

$$
\hat I=
V(D)
\frac{1}{N}
\sum_{i=1}^{N}
f(\mathbf{X}_i)
$$

对于积分区域：

$$
D=[a_1,b_1]\times\cdots\times[a_d,b_d]
$$

区域体积为：

$$
V(D)=
\prod_{j=1}^{d}(b_j-a_j)
$$

在每一个维度上随机生成：

$$
X_{ij}\sim U(a_j,b_j)
$$

其中：

$$
i=1,2,\dots,N
$$

$$
j=1,2,\dots,d
$$

则高维积分：

$$
I=\int_D f(\mathbf{x})\,d\mathbf{x}
$$

可以近似为：

$$
\hat I=
\left[
\prod_{j=1}^{d}(b_j-a_j)
\right]
\frac{1}{N}
\sum_{i=1}^{N}
f(X_{i1},X_{i2},\dots,X_{id})
$$

## 为什么适合高维积分

传统网格积分方法中，假设每个维度设置 $m$ 个网格点。对于 $d$ 维积分，总网格点数为：

$$
m^d
$$

例如每个维度设置 10 个点：

- 2 维时：

$$
10^2=100
$$

- 5 维时：

$$
10^5=100000
$$

- 10 维时：

$$
10^{10}
$$

可以看到，随着维数增加，计算量会迅速增长。

而蒙特卡洛方法只需要指定随机样本数量 $N$。

例如：

$$
N=100000
$$

无论是 5 维、10 维还是 20 维，都可以使用这 $100000$ 个随机样本进行积分估计。

因此蒙特卡洛方法特别适合处理高维积分问题。

高维定积分的随机模拟可以简单理解为：

**在高维积分区域中随机生成大量样本点，计算这些随机点上的函数值，然后利用函数值的平均值乘以积分区域的体积来估计积分。**

核心公式为：

$$
\boxed{
\text{积分估计值}
=
\text{积分区域体积}
\times
\text{随机样本函数值的平均值}
}
$$

即：

$$
\boxed{
\hat I=
V(D)
\frac{1}{N}
\sum_{i=1}^{N}
f(\mathbf{X}_i)
}
$$

## 4. 手动实现

```python
N = 10000
np.random.seed(42)
d = 3
a = np.array([0, 0, 0])
b = np.array([1, 1, 1])

X = np.random.uniform(a, b, size=(N, d))
f = X[:, 0]**2 + X[:, 1]**2 + X[:, 2]**2
V = np.prod(b - a)
I = V * np.mean(f)

print(I)
```
