# 02 平均值法 Mean value method

## 1. 算法介绍



随机模拟积分平均值法，也叫 **蒙特卡洛积分平均值法**，是一种利用随机抽样来近似计算定积分的方法。

假设要求计算定积分：

$$
I=\int_a^b f(x)\,dx
$$

基本思想是：

在区间 $[a,b]$ 上随机生成大量服从均匀分布的随机数：

$$
X_1,X_2,\cdots,X_N
$$

计算这些随机点对应的函数值：

$$
f(X_1),f(X_2),\cdots,f(X_N)
$$

然后计算函数值的平均值：

$$
\overline{f}
=
\frac{1}{N}\sum_{i=1}^{N}f(X_i)
$$

最后用区间长度乘以函数值的平均值：

$$
I\approx (b-a)\overline{f}
$$

因此，随机模拟积分平均值法可以简单理解为：

**随机取点 → 计算函数值 → 求平均值 → 乘以区间长度**


## 2. 数学原理



设随机变量 $X$ 在区间 $[a,b]$ 上服从均匀分布：

$$
X\sim U(a,b)
$$

其概率密度函数为：

$$
p(x)=\frac{1}{b-a},\quad a\leq x\leq b
$$

根据数学期望的定义：

$$
E[f(X)]
=
\int_a^b f(x)p(x)\,dx
$$

代入均匀分布的概率密度函数：

$$
E[f(X)]
=
\int_a^b f(x)\frac{1}{b-a}\,dx
$$

因此：

$$
E[f(X)]
=
\frac{1}{b-a}\int_a^b f(x)\,dx
$$

两边同时乘以 $b-a$，得到：

$$
\int_a^b f(x)\,dx
=
(b-a)E[f(X)]
$$

这就是随机模拟积分平均值法的数学基础。

实际计算中，数学期望 $E[f(X)]$ 一般无法直接得到。

因此，在区间 $[a,b]$ 上随机抽取 $N$ 个样本：

$$
X_1,X_2,\cdots,X_N
$$

根据大数定律，当 $N$ 足够大时：

$$
\frac{1}{N}\sum_{i=1}^{N}f(X_i)
\approx
E[f(X)]
$$

所以：

$$
\int_a^b f(x)\,dx
\approx
(b-a)
\frac{1}{N}
\sum_{i=1}^{N}f(X_i)
$$

因此，随机模拟积分平均值法的估计公式为：

$$
I_N=
\frac{b-a}{N}
\sum_{i=1}^{N}f(X_i)
$$

其中：

- $N$：随机抽样次数
- $X_i$：区间 $[a,b]$ 上生成的随机数
- $f(X_i)$：随机点对应的函数值
- $I_N$：定积分的近似值

对于定积分：

$$
I=\int_a^b f(x)\,dx
$$

随机模拟积分平均值法的步骤如下：

1. 确定被积函数 $f(x)$ 和积分区间 $[a,b]$。
2. 在区间 $[a,b]$ 上生成 $N$ 个均匀分布的随机数。
3. 计算每个随机数对应的函数值 $f(X_i)$。
4. 求所有函数值的平均值：

$$
\overline{f}
=
\frac{1}{N}\sum_{i=1}^{N}f(X_i)
$$

5. 用区间长度乘以函数值平均值：

$$
I_N=(b-a)\overline{f}
$$

最终得到定积分的近似值。

## 4. 手动实现

```python
N = 10000
a = 3
b = 5
x = np.random.uniform(a, b, N)
fx = x**2
mean_fx = np.mean(fx)
I = (b-a)*mean_fx

print(I)
```
